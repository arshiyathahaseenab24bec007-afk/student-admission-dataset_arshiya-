```python
#select the features and targets

# Feature columns relevant for admission probability prediction
X = df.drop(['student_id', 'admission_probability'], axis=1)

# Target column
y = df['admission_probability']
```

```python
#Encode categorical data(linear regression cannot handle text)

X_encoded = pd.get_dummies(X, drop_first=True)

X_encoded.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)
print("Shape of X_train after encoding and splitting:", X_train.shape)
print("Shape of X_test after encoding and splitting:", X_test.shape)
```
